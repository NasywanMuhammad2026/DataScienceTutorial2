## 5.Create two random numpy array of size n x p and p x m and then compute the dot product

```python
arsyadkebab = np.random.rand(2, 3)
rand_arr_trans = np.random.rand(3, 6)
dot_product = np.dot(arsyadkebab, rand_arr_trans)

print("Array 1 (2x3):\n", arsyadkebab)
print("\nArray 2 (3x6):\n", rand_arr_trans)
print("\nDot Product (2x6):\n", dot_product)
```
